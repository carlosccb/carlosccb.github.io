__Load libraries__

```python
import pandas as pd
import numpy as np

import sklearn

import seaborn as sns
import matplotlib.pyplot as plt
```

__Create artificial data with sklearn__

```python
from sklearn.datasets import make_classification

data = make_classification(n_samples=100, n_features=2, n_informative=2,
                           n_redundant=0, n_clusters_per_class=1,
                           class_sep=1, n_classes=3)
```

Doesn't look too promising...

```python
%matplotlib inline
sns.scatterplot(x=data[0][:,0], y=data[0][:,1], hue=data[1]);
```

__Load Iris__

Load a famous dataset instead to see if the data looks better

```python
from sklearn.datasets import load_iris

iris_data = load_iris()
```

```python
iris_data.keys()
```

```python
{i: iris_data["feature_names"][i] for i in range(4)}
```

```python
iris_data = {i:iris_data[i] for i in ["data", "target", "feature_names"]}
```

Variables 0 and 2 look good

```python
sns.pairplot(pd.DataFrame(iris_data["data"]));
```

```python
iris_df = pd.DataFrame({"x": iris_data["data"][:,2],
                        "y": iris_data["data"][:,0],
                        "target": iris_data["target"]})
```

It does look good!

```python
iris_df.plot.scatter(x="x", y="y", c="target", colormap='viridis');
```

Save variables to csv

```python
iris_df.to_csv("iris_class_data.csv", index=False)
```
